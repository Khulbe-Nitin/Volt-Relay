```python
# Convert timestamp
swap_events["event_ts"] = pd.to_datetime(
    swap_events["event_ts"],
    errors="coerce"
)

print("Missing timestamps:", swap_events["event_ts"].isna().sum())

print("Minimum timestamp:", swap_events["event_ts"].min())
print("Maximum timestamp:", swap_events["event_ts"].max())
```

```python
#Timestamp affected records
bug_start = pd.Timestamp("2025-03-10 00:00:00")
bug_end = pd.Timestamp("2025-04-14 23:59:59")

affected_mask = (
    (swap_events["station_firmware"] == "v3.2.0") &
    (swap_events["event_ts"] >= bug_start) &
    (swap_events["event_ts"] <= bug_end)
)

print("Affected records:", affected_mask.sum())

print("\nAffected records by firmware:")
print(swap_events.loc[affected_mask, "station_firmware"].value_counts())

print("\nAffected records by sync mode:")
print(swap_events.loc[affected_mask, "sync_mode"].value_counts())

print("\nAffected timestamp range:")
print(swap_events.loc[affected_mask, "event_ts"].min())
print(swap_events.loc[affected_mask, "event_ts"].max())
```

```python
swap_events["event_ts_original"] = swap_events["event_ts"]

swap_events["event_ts_corrected"] = swap_events["event_ts"]

swap_events.loc[affected_mask, "event_ts_corrected"] = (
    swap_events.loc[affected_mask, "event_ts"] +
    pd.Timedelta(hours=5, minutes=30)
)
```

```python
print("Original minimum:", swap_events["event_ts_original"].min())
print("Corrected minimum:", swap_events["event_ts_corrected"].min())

print("\nOriginal maximum:", swap_events["event_ts_original"].max())
print("Corrected maximum:", swap_events["event_ts_corrected"].max())

print("\nCorrections made:",
      (swap_events["event_ts_original"] != swap_events["event_ts_corrected"]).sum())
```

```python
#checking Duplicate Rows
print("Exact duplicate rows:", swap_events.duplicated().sum())

print("Duplicate event IDs:", swap_events["event_id"].duplicated().sum())
```

```python
#Checking duplication by Column
same_rider_station_time = (
    swap_events
    .groupby(["rider_id", "station_id", "event_ts_original"])
    .size()
    .reset_index(name="count")
)

print("Groups with multiple events:",
      (same_rider_station_time["count"] > 1).sum())

print("\nMaximum events in one rider-station-timestamp group:",
      same_rider_station_time["count"].max())
```

```python
duplicate_groups = same_rider_station_time[
    same_rider_station_time["count"] > 1
]

print(duplicate_groups.head(20))
```

```python
# Sort by rider, station and corrected timestamp
swap_events = swap_events.sort_values(
    ["rider_id", "station_id", "event_ts_corrected"]
).reset_index(drop=True)

# Time difference from the previous event for the same rider at the same station
swap_events["prev_event_ts"] = (
    swap_events
    .groupby(["rider_id", "station_id"])["event_ts_corrected"]
    .shift(1)
)

swap_events["seconds_since_prev"] = (
    swap_events["event_ts_corrected"] -
    swap_events["prev_event_ts"]
).dt.total_seconds()

print(swap_events["seconds_since_prev"].describe())
```

```python
print("Events within 1 minute:",
      (swap_events["seconds_since_prev"] <= 60).sum())

print("Events within 5 minutes:",
      (swap_events["seconds_since_prev"] <= 300).sum())

print("Events within 15 minutes:",
      (swap_events["seconds_since_prev"] <= 900).sum())
```

```python
sns.set_style("whitegrid")

stations["is_test_station"] = stations["station_id"].str.startswith("STN-TST")
events = swap_events.merge(
    stations[["station_id","city","charger_generation","is_test_station","grid_tariff_inr_kwh"]],
    on="station_id", how="left"
)
events_prod = events[~events["is_test_station"]].copy()
completed_events = events_prod[events_prod["is_completed"]].copy()

monthly = events_prod.groupby("year_month").agg(
    total_attempts=("event_id","count"),
    failed_attempts=("is_failed","sum"),
).reset_index()
monthly_rev = completed_events.groupby("year_month").agg(
    completed_swaps=("event_id","count"),
    revenue_inr=("revenue_inr","sum"),
).reset_index()
monthly = monthly.merge(monthly_rev, on="year_month")
monthly["failure_rate_pct"] = monthly["failed_attempts"]/monthly["total_attempts"]*100

fig, axes = plt.subplots(2,2, figsize=(14,8))
sns.lineplot(data=monthly, x="year_month", y="completed_swaps", marker="o", ax=axes[0,0], color="#2E86AB")
sns.lineplot(data=monthly, x="year_month", y="revenue_inr", marker="o", ax=axes[0,1], color="#06A77D")
sns.lineplot(data=monthly, x="year_month", y="failure_rate_pct", marker="o", ax=axes[1,0], color="#E63946")
for ax in axes.flat: ax.tick_params(axis="x", rotation=60)
plt.tight_layout(); plt.show()
monthly
```

<h1> Service failure and Customer Experiece </h1>

```python
events_prod = events_prod.merge(riders[["rider_id","vehicle_class"]], on="rider_id", how="left")

dow_order = ["Monday","Tuesday","Wednesday","Thursday","Friday","Saturday","Sunday"]
heat_failure = (
    events_prod.pivot_table(index="day_of_week", columns="event_hour", values="is_failed", aggfunc="mean")
    .reindex(dow_order) * 100
)

plt.figure(figsize=(14,5))
sns.heatmap(heat_failure, cmap="rocket_r", cbar_kws={"label":"Failure rate %"})
plt.title("Failure Rate by Hour of Day x Day of Week")
plt.tight_layout(); plt.show()

station_perf = events_prod.groupby("station_id").agg(
    failure_rate_pct=("is_failed","mean"), attempts=("event_id","count")
).reset_index()
station_perf["failure_rate_pct"] *= 100
print(station_perf.sort_values("failure_rate_pct", ascending=False).head(10))

vc = events_prod.groupby("vehicle_class")["is_failed"].mean()*100
print(vc)
```

Failure rate is fairly uniform across the week (~2.8%–4.5%) — no dramatic single-hour spike, but the darkest cells cluster in late-night/overnight hours (19:00–23:00 and 0:00–5:00) across most days, while weekday daytime (8am–5pm) tends to run lower. <br> <br> <br>
Likely story: overnight, charged-battery stock depletes after evening demand and restocking/monitoring is slower — worth checking against station_hourly_status.charged_2w_avg by hour if you want to confirm.
3 of your top 4 worst stations are in Delhi (STN-DEL-037, STN-DEL-048, STN-DEL-045) plus one in Jaipur. That's not "a few bad stations scattered randomly" — it's a city-level concentration, which points to something systemic in Delhi (grid outages, heat, demand overload) rather than isolated hardware faults. That's worth its own chart.

```python
# 1. Which cities actually drive the failures?
city_fail = events_prod.groupby("city")["is_failed"].mean().sort_values(ascending=False) * 100
plt.figure(figsize=(8,5))
sns.barplot(x=city_fail.values, y=city_fail.index, hue=city_fail.index, palette="rocket", legend=False)
plt.title("Failure Rate % by City")
plt.xlabel("Failure rate (%)")
plt.tight_layout(); plt.show()

# 2. Worst 15 stations by queue wait (not just failure) + how failure is distributed network-wide
fig, axes = plt.subplots(1,2, figsize=(14,5))
worst_wait = events_prod.groupby("station_id")["queue_wait_min"].mean().sort_values(ascending=False).head(15)
sns.barplot(x=worst_wait.values, y=worst_wait.index, hue=worst_wait.index, palette="mako", ax=axes[0], legend=False)
axes[0].set_title("Top 15 Stations by Avg Queue Wait")

sns.histplot(station_perf["failure_rate_pct"], bins=25, kde=True, color="#E63946", ax=axes[1])
axes[1].axvline(station_perf["failure_rate_pct"].mean(), color="black", linestyle="--", label="network avg")
axes[1].set_title("Distribution of Failure Rate Across All Stations")
axes[1].legend()
plt.tight_layout(); plt.show()

# 3. Vehicle class + abandonment check
print(events_prod.groupby("vehicle_class")["is_failed"].mean()*100)
print(events_prod.groupby("vehicle_class")["is_abandoned"].mean()*100)
print(events_prod.groupby("vehicle_class")["queue_wait_min"].mean())
```

The failure-rate distribution is bimodal — most stations cluster around 2.5–3.5% (near the network average, dashed line), but there's a second small bump around 5–6%. That's not a smooth long tail; it's a distinct group of consistently bad stations sharing some common trait. Worth segmenting those ~10-15 outlier stations by city/charger-generation/connectivity to find what they have in common — that's a strong "anomalies" callout for your report.
Queue wait is a different problem than failure — the worst-wait stations are spread across Delhi, Bengaluru, Jaipur, Mumbai, Pune, Hyderabad fairly evenly, unlike failure rate which concentrated in specific cities. So high failure ≠ high wait; they likely have different root causes.
